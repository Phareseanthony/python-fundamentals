```python

marks = []

for i in range(5):
    mark = float(input(f"Enter marks {i + 1}: "))
    marks.append(mark)

print("All marks:", marks)
print("The highest mark:", max(marks))
print("The lowest mark:", min(marks))

average = sum(marks) / len(marks)

print("The average mark:", average)
```

```python

# STEP 1: Create empty lists
students = []
marks = []
averages = []
grades = []

# STEP 2: Collect information for three students
for i in range(3):

    name = input(f"Enter student {i + 1} name: ")
    students.append(name)

    student_marks = []

    # Collect three subject marks for each student
    for j in range(3):
        mark = float(input(f"Enter mark for subject {j + 1}: "))
        student_marks.append(mark)

    marks.append(student_marks)

    # STEP 3: Calculate average
    average = sum(student_marks) / len(student_marks)
    averages.append(average)

    # STEP 4: Determine grade
    if average >= 80:
        grade = "A"
    elif average >= 70:
        grade = "B"
    elif average >= 60:
        grade = "C"
    elif average >= 50:
        grade = "D"
    else:
        grade = "E"

    grades.append(grade)


# STEP 5: Identify the student with the highest average
highest_average = max(averages)
highest_index = averages.index(highest_average)
top_student = students[highest_index]


# STEP 6: Display all student results
print("\n========== STUDENT RESULTS ==========")

for i in range(3):
    print(f"\nStudent Name: {students[i]}")
    print(f"Marks: {marks[i]}")
    print(f"Average: {averages[i]:.2f}")
    print(f"Grade: {grades[i]}")

print("\n========== TOP STUDENT ==========")
print("Student:", top_student)
print("Highest Average:", highest_average)
```

```python
student = {
    "name": "Pharese",
    "age": 21,
    "course": "Computer Science",
    "year": "Second Year"
}

# Display keys
print(student.keys())

# Display values
print(student.values())

# Display key-value pairs
print(student.items())

# Update dictionary
student.update({
    "salary": 40000,
    "status": "Permanent"
})

print(student)

# Remove last inserted item
removed_item = student.popitem()

print("Removed item:", removed_item)
print("Final dictionary:", student)
```
